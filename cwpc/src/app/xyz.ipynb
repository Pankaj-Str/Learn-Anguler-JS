```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

print(sns.__version__)
```

```python
sns.set_theme(style="whitegrid",palette="deep")
plt.rcParams['figure.figsize'] = (8, 5)  
```

```python
tips = sns.load_dataset("tips")
tips.head()
tips.tail()
```

```python
print(tips.shape)
```

```python
print(tips.info)
```

```python
sns.scatterplot(data=tips,
                x="total_bill",
                y="tip",
                hue="sex",
                style="smoker",
                palette="deep")
plt.title("Relational Plots (Relationship between two numeric variables)")
plt.show()
```

```python
data = sns.load_dataset("flights")
data.head()
```

```python
sns.lineplot(data=data,x="year",y="passengers",hue="month")
plt.show()
```

```python
import numpy as np
sns.barplot(data=tips, x="day", y="total_bill", hue="sex", estimator=np.mean)
plt.title("Average Total Bill by Day and Sex")
plt.show()
```
